# DenseNet121 – Oral Cancer Classification


```python
import os
import json
import random
from glob import glob
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
from IPython.display import display

from sklearn.metrics import (
    ConfusionMatrixDisplay,
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
from sklearn.utils.class_weight import compute_class_weight

import tensorflow as tf
from tensorflow.keras.applications.densenet import DenseNet121, preprocess_input
from tensorflow.keras.callbacks import (
    EarlyStopping,
    LearningRateScheduler,
    ModelCheckpoint
)
from tensorflow.keras.layers import (
    BatchNormalization,
    Dense,
    Dropout,
    GlobalAveragePooling2D,
    Input
)
from tensorflow.keras.metrics import Precision, Recall
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("TensorFlow version:", tf.__version__)
```

## 1. Common configuration

```python
SEED = 23
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

OUTPUT_PATH = "Data_subsets"
TRAIN_PATH = os.path.join(OUTPUT_PATH, "train")
VAL_PATH = os.path.join(OUTPUT_PATH, "val")
TEST_PATH = os.path.join(OUTPUT_PATH, "test")

IMG_DIMS = 100
TARGET_SIZE = (IMG_DIMS, IMG_DIMS)
COLOR_MODE = "rgb"
CLASS_MODE = "categorical"
BATCH_SIZE = 16
EPOCHS = 7
ETA = 0.0001

MODEL_NAME = "DenseNet121"
SAVE_STEM = "densenet121"
BEST_MODEL_PATH = f"{SAVE_STEM}_best.keras"

for p in [TRAIN_PATH, VAL_PATH, TEST_PATH]:
    if not os.path.isdir(p):
        raise FileNotFoundError(
            f"Dataset folder not found: {p}\n"
            "Expected Data_subsets/train, Data_subsets/val and Data_subsets/test."
        )

print("Model:", MODEL_NAME)
print("Train:", TRAIN_PATH)
print("Validation:", VAL_PATH)
print("Test:", TEST_PATH)
```

## 2. Dataset inspection

```python
class_names = sorted([
    d for d in os.listdir(TRAIN_PATH)
    if os.path.isdir(os.path.join(TRAIN_PATH, d))
])
print("Classes:", class_names)

counts = {}
for cls in class_names:
    counts[cls] = len([
        f for f in os.listdir(os.path.join(TRAIN_PATH, cls))
        if os.path.isfile(os.path.join(TRAIN_PATH, cls, f))
    ])

print("Training image counts:", counts)

plt.figure(figsize=(7, 4))
plt.bar(list(counts.keys()), list(counts.values()))
plt.title("Training Images per Class")
plt.ylabel("Number of Images")
plt.tight_layout()
plt.show()
```

## 3. Data generators

```python
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=5,
    width_shift_range=0.05,
    height_shift_range=0.05,
    brightness_range=[0.9, 1.1],
    zoom_range=[0.9, 1.1],
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)
test_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

train_gen = train_datagen.flow_from_directory(
    TRAIN_PATH,
    target_size=TARGET_SIZE,
    color_mode=COLOR_MODE,
    class_mode=CLASS_MODE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED
)

val_gen = val_datagen.flow_from_directory(
    VAL_PATH,
    target_size=TARGET_SIZE,
    color_mode=COLOR_MODE,
    class_mode=CLASS_MODE,
    batch_size=BATCH_SIZE,
    shuffle=False,
    seed=SEED
)

test_gen = test_datagen.flow_from_directory(
    TEST_PATH,
    target_size=TARGET_SIZE,
    color_mode=COLOR_MODE,
    class_mode=CLASS_MODE,
    batch_size=BATCH_SIZE,
    shuffle=False,
    seed=SEED
)

print("Class indices:", train_gen.class_indices)
```

## 4. Build DenseNet121

```python
base_model = DenseNet121(
    include_top=False,
    weights="imagenet",
    input_tensor=Input(shape=(IMG_DIMS, IMG_DIMS, 3)),
    input_shape=(IMG_DIMS, IMG_DIMS, 3)
)

for layer in base_model.layers:
    layer.trainable = False

num_classes = train_gen.num_classes

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    BatchNormalization(),
    Dense(512, activation="relu"),
    Dropout(0.4),
    BatchNormalization(),
    Dense(256, activation="relu"),
    Dropout(0.3),
    Dense(num_classes, activation="softmax")
])

model.summary()
```

## 5. Compile and train

```python
model.compile(
    optimizer=Adam(learning_rate=ETA),
    loss="categorical_crossentropy",
    metrics=[
        "categorical_accuracy",
        Precision(name="precision"),
        Recall(name="recall")
    ]
)

cls_wt = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_gen.classes),
    y=train_gen.classes
)
class_weight = {i: float(w) for i, w in enumerate(cls_wt)}
print("Class weights:", class_weight)

def scheduler(epoch, lr):
    return ETA * 0.1 if epoch > 19 else ETA

callbacks = [
    LearningRateScheduler(scheduler),
    ModelCheckpoint(
        BEST_MODEL_PATH,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,
        verbose=1
    )
]

history = model.fit(
    train_gen,
    epochs=EPOCHS,
    validation_data=val_gen,
    class_weight=class_weight,
    callbacks=callbacks
)

pd.DataFrame(history.history).to_csv(
    f"{SAVE_STEM}_training_history.csv",
    index=False
)
```

## 6. Training curves

```python
def plot_history(train_key, val_key, ylabel, filename):
    if train_key not in history.history or val_key not in history.history:
        return
    plt.figure(figsize=(8, 5))
    plt.plot(history.history[train_key], label="Training")
    plt.plot(history.history[val_key], label="Validation")
    plt.xlabel("Epoch")
    plt.ylabel(ylabel)
    plt.title(f"{MODEL_NAME} - {ylabel}")
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=200)
    plt.show()

plot_history("loss", "val_loss", "Loss", f"{SAVE_STEM}_loss.png")
plot_history(
    "categorical_accuracy",
    "val_categorical_accuracy",
    "Categorical Accuracy",
    f"{SAVE_STEM}_accuracy.png"
)
```

## 7. Test evaluation

```python
best_model = load_model(BEST_MODEL_PATH)

test_gen.reset()
probabilities = best_model.predict(test_gen, verbose=1)
y_true = test_gen.classes
y_pred = np.argmax(probabilities, axis=1)

idx_to_class = {v: k for k, v in test_gen.class_indices.items()}
ordered_labels = list(range(len(idx_to_class)))
ordered_names = [idx_to_class[i] for i in ordered_labels]

report_dict = classification_report(
    y_true,
    y_pred,
    labels=ordered_labels,
    target_names=ordered_names,
    output_dict=True,
    zero_division=0
)
report_df = pd.DataFrame(report_dict).transpose()
display(report_df)
report_df.to_csv(f"{SAVE_STEM}_classification_report.csv")

accuracy = accuracy_score(y_true, y_pred)
macro_precision = precision_score(y_true, y_pred, average="macro", zero_division=0)
macro_recall = recall_score(y_true, y_pred, average="macro", zero_division=0)
macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)

print(f"Accuracy       : {accuracy:.4f}")
print(f"Macro Precision: {macro_precision:.4f}")
print(f"Macro Recall   : {macro_recall:.4f}")
print(f"Macro F1-score : {macro_f1:.4f}")
```

## 8. Confusion matrix

```python
cm = confusion_matrix(y_true, y_pred, labels=ordered_labels)

fig, ax = plt.subplots(figsize=(7, 6))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=ordered_names
)
disp.plot(ax=ax, values_format="d")
plt.title(f"{MODEL_NAME} - Confusion Matrix")
plt.tight_layout()
plt.savefig(f"{SAVE_STEM}_confusion_matrix.png", dpi=200)
plt.show()
```

## 9. Save standardized comparison results

```python
results = {
    "model": MODEL_NAME,
    "image_size": [IMG_DIMS, IMG_DIMS],
    "batch_size": BATCH_SIZE,
    "epochs_requested": EPOCHS,
    "epochs_completed": len(history.history.get("loss", [])),
    "class_indices": train_gen.class_indices,
    "test_samples": int(test_gen.samples),
    "accuracy": float(accuracy),
    "macro_precision": float(macro_precision),
    "macro_recall": float(macro_recall),
    "macro_f1": float(macro_f1),
    "best_model_path": BEST_MODEL_PATH
}

with open(f"{SAVE_STEM}_results.json", "w") as f:
    json.dump(results, f, indent=2)

print(json.dumps(results, indent=2))
```

## 10. Single-image prediction

```python
CLASS_NAMES = {v: k for k, v in train_gen.class_indices.items()}

def predict_image(image_path, show=True):
    img = tf.keras.utils.load_img(image_path, target_size=TARGET_SIZE)
    x = tf.keras.utils.img_to_array(img)
    x = np.expand_dims(x, axis=0)

    # Correct preprocessing for DenseNet121.
    x = preprocess_input(x)

    probs = best_model.predict(x, verbose=0)[0]
    pred_index = int(np.argmax(probs))
    pred_class = CLASS_NAMES[pred_index]
    confidence = float(probs[pred_index]) * 100.0

    if show:
        plt.figure(figsize=(5, 5))
        plt.imshow(img)
        plt.axis("off")
        plt.title(f"{MODEL_NAME}: {pred_class} ({confidence:.2f}%)")
        plt.show()

    print("Prediction :", pred_class)
    print("Confidence :", f"{confidence:.2f}%")
    return pred_class, confidence, probs
```
